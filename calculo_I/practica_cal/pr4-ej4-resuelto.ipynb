### Ejercicio 4

Determina los extremos relativos y absolutos (si existen) de la función $f(x)=|x^2-16|$.

1. Escribe la función $f$ como una función definida a trozos.
2. Estudia la derivabilidad de la función $f$ en los puntos en los que cambia su definición.
3. Determina el conjunto de posibles extremos de la función $f$.
4. Estudia las propiedades de monotonía de la función $f$. ¿Tiene $f$ extremos relativos? En caso afirmativo, ¿son máximos o mínimos? ¿Tiene $f$ extremos absolutos?
5. ¿Tiene la función $f$ extremos absolutos en el intervalo $[-4,5]$? Justifica tu respuesta y, si es afirmativa, determina los extremos absolutos de $f$ en dicho intervalo.
6. Representa gráficamente la función, señalando en la gráfica sus extremos.


```python
# APARTADO 1

import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# Determinamos dónde se anula f, para ver dónde puede cambiar de signo el argumento del valor absoluto:
x = sp.Symbol('x',real=True)
f = sp.Lambda((x),sp.Abs(x**2-16))
display('Donde se anula la función f: ', sp.solve(f(x),x))
```

```python
# Evaluamos x**2-16 en un punto del intervalo (-oo,-4):
g = sp.Lambda((x),x**2-16)
g(-5)
```

```python
# Repetimos el proceso con un punto en (-4,4) y otro en (4,oo):
display('g(0)=',g(0))
display('g(5)=',g(5))
```

```python
# Concluimos que, al ser la función continua, el argumento del valor absoluto es negativo en (-4,4) y positivo en el resto
# Representamos gráficamente f en el intervalo [-10,10]:
xx = np.linspace(-10,10,100)
ff = np.piecewise(xx,[(xx<-4)|(xx>4),(xx>=-4) & (xx<=4)], [lambda xx: xx**2-16, lambda xx: 16-xx**2])
plt.figure()
plt.plot(xx,ff,label='y=f(x)')
plt.title('Gráfica de f')
plt.legend()
```

```python
# APARTADO 2

# Los puntos en los que cambia la definición de f son x=4 y x=-4
x = sp.Symbol('x',real=True)
f = sp.Lambda((x),sp.Abs(x**2-16))

# Derivabilidad en x = 4:
cociente = (f(x)-f(4))/(x-4) # Cociente incremental
dfdcha = sp.limit(cociente,x,4,'+') # Derivada en 4 por la derecha
dfizqda = sp.limit(cociente,x,4,'-') # Derivada en 4 por la izquierda

display('Derivada en a=4 por la derecha = ',dfdcha)
display('Derivada en a=4 por la izquierda = ',dfizqda)


# Derivabilidad en x = -4:
cociente = (f(x)-f(-4))/(x+4) # Cociente incremental
dfdcha = sp.limit(cociente,x,-4,'+') # Derivada en -4 por la derecha
dfizqda = sp.limit(cociente,x,-4,'-') # Derivada en -4 por la izquierda

display('Derivada en a=-4 por la derecha = ',dfdcha)
display('Derivada en a=-4 por la izquierda = ',dfizqda)

# Como las derivadas laterales son distintas, f no es derivable en 4 ni en -4
# Como la función es par, en realidad, es suficiente estudiar qué ocurre en uno de los puntos
```

```python
# APARTADO 3

# Los posibles extremos de la función f en R son:
# - Los puntos de no derivabilidad: -4, 4
# - Los puntos en los que se anula la primera derivada:

dfpos = sp.diff(x**2-16,x)
display('Derivada en x>4 o x<-4 es ',dfpos)
display('Ceros de la derivada cuando x>4 o x<-4:',sp.solve(dfpos,x)) # Está fuera del conjunto considerado

dfneg = sp.diff(16-x**2,x)
display('Derivada entre -4 y 4 es ',dfneg)
display('Ceros de la derivada entre -4 y 4:',sp.solve(dfneg,x))

# Así, los posibles extremos relativos son -4, 4 y 0 (único punto donde se anula la primera derivada) 
```

```python
# APARTADO 4

# Dado que hay puntos de no derivabilidad, para estudiar la monotonía de f, empleamos el criterio de la primera derivada.
# Evaluamos f' en un punto en (-oo,-4), (-4,0), (0,4) y (4,+oo):
display('Valor de la derivada de f en -5 = ',dfpos.subs(x,-5)) # -10<0 -> f decreciente en (-oo,-4)
display('Valor de la derivada de f en -2 = ',dfneg.subs(x,-2)) # 4>0 -> f creciente en (-4,0)
display('Valor de la derivada de f en 2 = ',dfneg.subs(x,2)) # -4>0 -> f decreciente en (0,4)
display('Valor de la derivada de f en 5 = ',dfpos.subs(x,5)) # 10>0 -> f creciente en (4,+oo)
# Como consecuencia, al ser f continua, f tiene dos mínimos locales en -4 y 4 y un máximo local en 0.
```

```python
# Para ver si f tiene extremos absolutos en R, calculamos los límites en +oo (y -oo)
# En realidad, como la función es par, es suficiente calcular uno de ellos:
from sympy import oo
l1 = sp.limit(f(x),x,-oo)
l2 = sp.limit(f(x),x,oo)
display('Limite en -oo = ',l1)
display('Limite en oo = ',l2)
# Como ambos límites son +oo, 0 no es un máximo absoluto.
# Los puntos -4 y 4 son mínimos absolutos.
```

```python
# APARTADO 5

# Como la función es continua en el intervalo cerrado [-4,5], 
# por el Teorema de Weierstrass, tiene extremos absolutos en dicho intervalo.
#
# Los posibles extremos en [-4,5] son: 
# - Los extremos del intervalo: -4 y 5
# - Los puntos de no derivabilidad en el interior del intervalo: 4
# - Los puntos del interior del intervalo en los que se anula la derivada: 0
#
# Para determinar los extremos absolutos, evaluamos f en los posibles extremos:
display('f(-4) = ',f(-4)) # 0 Como la función es par, toma el mismo valor en 4 y en -4
display('f(5) = ',f(5)) # 9
display('f(4) = ',f(4)) # 0
display('f(0) = ', f(0)) # 16
# La función f alcanza el mayor valor en [-4,5] en x=0, donde vale 16, 
# y el menor valor en x=4 y x=-4, donde vale 0.
# Por tanto, 4 y -4 son mínimos absolutos, y 0 es un máximo absoluto.
```

```python
# APARTADO 6

# Representamos la función gráficamente en el intervalo [-4,5]:
xx = np.linspace(-4,5,100)
yy = np.abs(xx**2-16)

plt.figure()
plt.plot(xx,yy)

plt.plot([-4,0,4], [0,16,0],'ro', markersize=8)
```
