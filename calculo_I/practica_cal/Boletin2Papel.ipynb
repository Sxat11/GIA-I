```python
import sympy as sp

x, h, k, a = sp.symbols('x h k a', real=True)
a = 0
def f(x):
    return sp.Piecewise(
        (x**2 * sp.sin(1/x) + k*x, x != 0),
        (a, x == 0)
    )

cociente = (f(h) - a) / h
derivada = sp.limit(cociente, h, a)

print("El valor analítico de f'(0) calculado es:")
display(derivada)
```

```python
import sympy as sp

x = sp.symbols('x')

f = x**-8 + x**-2 - 5

f_prime = sp.diff(f, x)

print("La función original es:")
display(f)

print("\nLa derivada es:")
display(f_prime)
```

```python
import sympy as sp

x, h = sp.symbols('x h', real=True)

# 1. Función f(x) = |x - 1|
f = sp.Abs(x - 1)

# Derivadas laterales de f en x = 1 mediante cociente incremental
f_der_der = sp.limit((f.subs(x, 1 + h) - f.subs(x, 1)) / h, h, 0, dir='+')
f_der_izq = sp.limit((f.subs(x, 1 + h) - f.subs(x, 1)) / h, h, 0, dir='-')

print("--- Función f(x) = |x - 1| ---")
print(f"Derivada lateral por la derecha f'(1+): {f_der_der}")
print(f"Derivada lateral por la izquierda f'(1-): {f_der_izq}")
print("Conclusión: No es derivable porque las derivadas laterales son distintas.\n")

# 2. Función g(x) = floor(x) (Parte entera)
g = sp.floor(x)

g_der_der = sp.limit((g.subs(x, 1 + h) - g.subs(x, 1)) / h, h, 0, dir='+')
g_der_izq = sp.limit((g.subs(x, 1 + h) - g.subs(x, 1)) / h, h, 0, dir='-')

print("--- Función g(x) = E(x) ---")
print(f"Derivada lateral por la derecha g'(1+): {g_der_der}")
print(f"Derivada lateral por la izquierda g'(1-): {g_der_izq}")
print("Conclusión: No es derivable porque la función no es continua en x = 1.")
```
