```python
import sympy as sp

x, h, k, a = sp.symbols('x h k a', real=True)
a = 0
def f(x):
    return sp.Piecewise(
        (x**2 * sp.sin(1/x) + k*x, x != 0),
        (a, x == 0)
    )

cociente = (f(h) - a) / h
derivada = sp.limit(cociente, h, a)

print("El valor analítico de f'(0) calculado es:")
display(derivada)
```

```python
import sympy as sp

x, h = sp.symbols('x h', real=True)

# 1. Función f(x) = |x - 1|
f = sp.Abs(x - 1)

# Derivadas laterales de f en x = 1 mediante cociente incremental
f_der_der = sp.limit((f.subs(x, 1 + h) - f.subs(x, 1)) / h, h, 0, dir='+')
f_der_izq = sp.limit((f.subs(x, 1 + h) - f.subs(x, 1)) / h, h, 0, dir='-')

print("--- Función f(x) = |x - 1| ---")
print(f"Derivada lateral por la derecha f'(1+): {f_der_der}")
print(f"Derivada lateral por la izquierda f'(1-): {f_der_izq}")
print("Conclusión: No es derivable porque las derivadas laterales son distintas.\n")

# 2. Función g(x) = floor(x) (Parte entera)
g = sp.floor(x)

g_der_der = sp.limit((g.subs(x, 1 + h) - g.subs(x, 1)) / h, h, 0, dir='+')
g_der_izq = sp.limit((g.subs(x, 1 + h) - g.subs(x, 1)) / h, h, 0, dir='-')

print("--- Función g(x) = E(x) ---")
print(f"Derivada lateral por la derecha g'(1+): {g_der_der}")
print(f"Derivada lateral por la izquierda g'(1-): {g_der_izq}")
print("Conclusión: No es derivable porque la función no es continua en x = 1.")
```

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

x = sp.symbols('x', real=True)

f = sp.Abs(x) / sp.exp(x)

print("Función:")
display(f)

f_prime = sp.diff(f, x)
print("\nDerivada :")
display(f_prime)

puntos_criticos = sp.solve(f_prime, x)
print(f"\nPuntos críticos hallados por anulación de la derivada: {puntos_criticos}")

print(f"f(0) = {f.subs(x, 0)}")
for pc in puntos_criticos:
    print(f"f({pc}) = {f.subs(x, pc).evalf()}")


def f_num(x):
    return np.abs(x) / np.exp(x)

x_vals = np.linspace(-1, 4, 800)
y_vals = f_num(x_vals)

plt.figure(figsize=(8, 5))
plt.plot(x_vals, y_vals, color='blue', label='$f(x) = \\frac{|x|}{e^x}$')

plt.scatter([0], [0], color='red', zorder=5, label='Mínimo relativo: $(0, 0)$')
x_max = 1.0
y_max = 1.0 / np.exp(1.0)
plt.scatter([x_max], [y_max], color='green', zorder=5, label=f'Máximo relativo: $(1, 1/e) \\approx (1, {y_max:.3f})$')

plt.title('Extremos relativos de $f(x) = \\frac{|x|}{e^x}$')
plt.xlabel('$x$')
plt.ylabel('$f(x)$')
plt.axhline(0, color='black', linewidth=0.8, linestyle='--')
plt.axvline(0, color='black', linewidth=0.8, linestyle='--')
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend()
plt.show()
```
