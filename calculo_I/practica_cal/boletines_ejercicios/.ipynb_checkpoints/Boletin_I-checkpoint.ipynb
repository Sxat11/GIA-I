```python
import sympy as sp

# 1. Definir variable y función
x = sp.Symbol('x', real=True)
f = x**3 - 3*x 

# 2. Hallar la derivada y los puntos críticos
derivada = sp.diff(f, x)
puntos_criticos = sp.solve(derivada, x)
print("Puntos críticos:", puntos_criticos)

# 3. Evaluar el signo de la derivada en un punto de prueba de cada intervalo
# Imagina que tus puntos críticos son -1 y 1. Los intervalos son: (-inf, -1), (-1, 1), (1, inf)
puntos_prueba = [-2, 0, 2]  # Un valor que caiga dentro de cada intervalo

for p in puntos_prueba:
    signo = derivada.subs(x, p)
    print(f"Si x = {p}, f'(x) = {signo}")
```

```python
SIMETRÍAS
```

```python
import sympy as sp

# 1. Definir la variable simbólica y la función
x = sp.Symbol('x', real=True)
f = x**2 + 3  # Cambia esto por tu función (ej: x**3 o x**2 + 3)

# 2. Calcular f(-x) sustituyendo x por -x
f_menos_x = f.subs(x, -x)

# 3. Comprobar las condiciones de simetría
es_par = sp.simplify(f_menos_x - f) == 0
es_impar = sp.simplify(f_menos_x + f) == 0

# 4. Mostrar resultados
print("¿Es simétrica par (respecto al eje Y)?", es_par)
print("¿Es simétrica impar (respecto al origen)?", es_impar)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Definimos el rango de x abarcando varios periodos (por ejemplo, de -4*pi a 4*pi)
x = np.linspace(-4 * np.pi, 4 * np.pi, 1000)

# 2. Elegimos la función a analizar (por ejemplo, f1(x) = sin(3x))
# Puedes cambiarla por cualquier otra función del ejercicio
y = np.sin(3 * x)

# 3. Configuración de la gráfica
plt.figure(figsize=(10, 4))
plt.plot(x, y, label=r'$f(x) = \sin(3x)$', color='teal', linewidth=2)

plt.title('Análisis gráfico de la periodicidad', fontsize=14)
plt.xlabel('Eje X')
plt.ylabel('Eje Y')



plt.legend(fontsize=12)
plt.show()
```

```python
import sympy as sp

# 1. Definir la variable simbólica real
x = sp.Symbol('x', real=True)

# 2. Definir la función
f = sp.sin(3 * x)

# 3. Proponer un periodo genérico T o buscarlo analíticamente
# Sabemos por trigonometría que el seno tiene un periodo base de 2*pi afectado por el coeficiente de x (en este caso 3).
# Vamos a comprobar si T = 2*pi / 3 cumple la condición de periodicidad: f(x + T) = f(x)
T = 2 * sp.pi / 3

# Evaluamos f(x + T)
f_desplazada = f.subs(x, x + T)

# Simplificamos la diferencia
diferencia = sp.simplify(f_desplazada - f)

# Comprobamos si la diferencia es 0
es_periodica = (diferencia == 0)

print("Función analizada:", f)
print("¿Es periódica para T = 2*pi / 3?", es_periodica)
if es_periodica:
    print(f"El periodo de la función es: T = {T}")
```

```python
import sympy as sp

# 1. Definir la variable simbólica real
x = sp.Symbol('x', real=True)

# 2. Definir las funciones f(x) y g(x) según el enunciado
f = (6 * x) / (x**2 - 9)
g = sp.sqrt(3 * x)

# 3. Calcular la composición (f ∘ g)(x) sustituyendo x en f por la expresión de g
fog = f.subs(x, g)

# Simplificamos la expresión resultante
fog_simplificada = sp.simplify(fog)

print("La función compuesta (f ∘ g)(x) es:")
display(fog_simplificada)

# 4. Calcular el valor de la composición en un punto concreto, por ejemplo x = 12
punto = 12
resultado_en_punto = fog.subs(x, punto)

print(f"\nCalculando (f ∘ g)({punto}) de forma exacta:")
display(resultado_en_punto)
```

```python
CONTINUIDAD
```

```python
import sympy as sp

# Definir la variable simbólica
x = sp.Symbol('x', real=True)

# Definir la función a trozos con SymPy
f = sp.Piecewise(
    (sp.sqrt(x**2 - 1), x <= -1),
    (x**2 + 2*x + 1, (x > -1) & (x <= 10)),
    (121/10 * x, x > 10)
)

print("--- ESTUDIO DE CONTINUIDAD EN PUNTOS DE FRONTERA ---")

# Comprobación en x = -1
lim_izq_1 = sp.limit(f, x, -1, '-')
lim_der_1 = sp.limit(f, x, -1, '+')
f_val_1 = f.subs(x, -1)

print(f"\nPunto x = -1:")
print(f"Límite por la izquierda: {lim_izq_1}")
print(f"Límite por la derecha: {lim_der_1}")
print(f"Valor de la función f(-1): {f_val_1}")
if lim_izq_1 == lim_der_1 == f_val_1:
    print("-> La función es CONTINUA en x = -1")
else:
    print("-> La función es DISCONTINUA en x = -1")

# Comprobación en x = 10
lim_izq_2 = sp.limit(f, x, 10, '-')
lim_der_2 = sp.limit(f, x, 10, '+')
f_val_2 = f.subs(x, 10)

print(f"\nPunto x = 10:")
print(f"Límite por la izquierda: {lim_izq_2}")
print(f"Límite por la derecha: {lim_der_2}")
print(f"Valor de la función f(10): {f_val_2}")
if lim_izq_2 == lim_der_2 == f_val_2:
    print("-> La función es CONTINUA en x = 10")
else:
    print("-> La función es DISCONTINUA en x = 10")
```

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# Definir la variable simbólica
x = sp.Symbol('x', real=True)

# Definir la función a trozos con SymPy para la visualización y análisis general
f_sym = sp.Piecewise(
    (sp.sqrt(x**2 - 1), x <= -1),
    (x**2 + 2*x + 1, (x > -1) & (x <= 10)),
    (121/10 * x, x > 10)
)

print("--- ESTUDIO DE CONTINUIDAD EN PUNTOS DE FRONTERA ---")

# Comprobación en x = -1
# Evaluamos directamente con la expresión analítica de cada lado para evitar el fallo del algoritmo de Gruntz en Piecewise
lim_izq_1 = sp.limit(sp.sqrt(x**2 - 1), x, -1, '-')
lim_der_1 = sp.limit(x**2 + 2*x + 1, x, -1, '+')
f_val_1 = sp.sqrt((-1)**2 - 1)

print(f"\nPunto x = -1:")
print(f"Límite por la izquierda: {lim_izq_1}")
print(f"Límite por la derecha: {lim_der_1}")
print(f"Valor de la función f(-1): {f_val_1}")
if lim_izq_1 == lim_der_1 == f_val_1:
    print("-> La función es CONTINUA en x = -1")
else:
    print("-> La función es DISCONTINUA en x = -1")

# Comprobación en x = 10
lim_izq_2 = sp.limit(x**2 + 2*x + 1, x, 10, '-')
lim_der_2 = sp.limit((121/10) * x, x, 10, '+')
f_val_2 = (10)**2 + 2*(10) + 1

print(f"\nPunto x = 10:")
print(f"Límite por la izquierda: {lim_izq_2}")
print(f"Límite por la derecha: {lim_der_2}")
print(f"Valor de la función f(10): {f_val_2}")
if lim_izq_2 == lim_der_2 == f_val_2:
    print("-> La función es CONTINUA en x = 10")
else:
    print("-> La función es DISCONTINUA en x = 10")


# --- REPRESENTACIÓN GRÁFICA ---
%matplotlib inline

# Crear rangos para cada tramo de la función
x1 = np.linspace(-10, -1, 400)
x2 = np.linspace(-1, 10, 400)
x3 = np.linspace(10, 15, 400)

# Evaluar las funciones numéricamente para cada tramo
y1 = np.sqrt(x1**2 - 1)
y2 = x2**2 + 2*x2 + 1
y3 = (121 / 10) * x3

plt.figure(figsize=(9, 6))
plt.plot(x1, y1, color='blue', label=r'$f(x) = \sqrt{x^2 - 1}$  si $x \in (-\infty, -1]$')
plt.plot(x2, y2, color='green', label=r'$f(x) = x^2 + 2x + 1$  si $x \in (-1, 10]$')
plt.plot(x3, y3, color='purple', label=r'$f(x) = \frac{121}{10}x$  si $x \in (10, +\infty)$')

# Marcar los puntos de frontera analizados
plt.plot(-1, 0, 'ro', markersize=8, label='Frontera continua ($x = -1, y = 0$)')
plt.plot(10, 121, 'ro', markersize=8, label='Frontera continua ($x = 10, y = 121$)')

plt.title('Estudio de Continuidad de la Función a Trozos')
plt.xlabel('$x$')
plt.ylabel('$f(x)$')
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(loc='upper left')
plt.show()
```
