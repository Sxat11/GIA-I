```python
import numpy as np
import matplotlib.pyplot as mtl
def f1(x):
    y = 0.2*x*x - np.pi*x - 5
    return y

def f2(x):
    y = np.pi*(np.sin(x) + np.sin(5.0*x))
    return y


def f3(x):
    y = np.pi**2 * np.sin(x) * np.exp(-x/10.)
    return y

x = np.linspace(0,10*np.pi,500)
mtl.figure()
mtl.xlabel('Altura(h)')
mtl.ylabel('Longitud(m)')
p1, p2, p3 = mtl.plot(x, f1(x), x, f2(x), x, f3(x))
mtl.legend(('Función 1','Función 2','Función 3'))
```

```python
import numpy as np
import matplotlib.pyplot as plt

def f1(x):
    y = 0.2*x*x - np.pi*x - 5
    return y

def f2(x):
    y = np.pi*(np.sin(x) + np.sin(5.0*x))
    return y


def f3(x):
    y = np.pi**2 * np.sin(x) * np.exp(-x/10.)
    return y

x = np.linspace(0, 10*np.pi, 500) # Array de abscisas

plt.xlabel('Tiempo (s)')
plt.ylabel('Amplitud (cm)')
plt.title('Representacion de tres funciones')

p1, p2, p3 = plt.plot(x, f1(x), x, f2(x), x, f3(x))
plt.legend(('polinómica', 'trigonométrica', 'exponencial'))
```

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(-6,3)
ytrozos = np.piecewise(x, [x < -1, (x >= -1) & (x <= 2), x > 2], 
                       [lambda x: (-x**3 + np.sin(np.pi*x)), 
                        lambda x: (x**2 - 3), 
                        lambda x: (np.exp(x-1) - 10)])

plt.figure()
plt.plot(x,ytrozos)
plt.legend('Examen')
```

```python
import sympy as sp
x,y = sp.symbols('x,y', real=True)
expr = (x-3)*(x-3)**2*(y-2)
expr_long = sp.expand(expr)
print(expr_long)
sp.pprint(expr_long) # Imprimir de forma semejante a con lápiz y papel
print()
expr_short = sp.factor(expr)
print(expr_short) # Factorizar expresión
expr = -3+(x**2-6*x+9)/(x-3)
expr_simple = sp.simplify(expr) # Simplificar expresión
sp.pprint(expr)
print(expr_simple)
res = expr.subs({x:1, y:2}) # Sustitución de las variables simbólicas por constantes
print(res)
```

```python
import sympy as sp

# Definimos la variable simbólica
x = sp.symbols('x')

# 1. Definir la ecuación
# En SymPy, la función exponencial se escribe como sp.exp()
# Las ecuaciones se definen con sp.Eq() o igualando la expresión a cero: sp.exp(x + 1) - 5
ecuacion = sp.Eq(sp.exp(x + 1), 5)

# 2. Utilizar solve
# solve devuelve una lista con las soluciones exactas (en este caso, en función de logaritmos naturales)
solucion_exacta = sp.solve(ecuacion, x)
print("Solución exacta:", solucion_exacta)

# 3. Obtener la solución numérica mediante float
# Extraemos el primer (y único) elemento de la lista con [0] y lo convertimos a número decimal
solucion_numerica = float(solucion_exacta[0])
print("Solución numérica:", solucion_numerica)
```
