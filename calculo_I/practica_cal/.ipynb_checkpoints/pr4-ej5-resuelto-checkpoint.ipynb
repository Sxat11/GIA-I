### Ejercicio 5:

Considera la función $$f(x)=\left\{\begin{array}{ll}
-x^2+1,& \text{si }x\leq -1,\\
\sin(\pi x), & \text{si }-1< x< 1,\\
e^{x^2}, & \text{si }1< x.
\end{array}\right.$$

Determina los intervalos en los que $f$ es cóncava y convexa. ¿Tiene la función $f$ algún punto de inflexión?

```python
# Como la función está definida en tres tramos, estudiamos cada uno de ellos de forma separada:
import sympy as sp
x = sp.Symbol('x', real=True)

f1 = -x**2 +1
f2 = sp.sin(sp.pi*x)
f3 = sp.exp(x**2)

# Al ser todas las funciones suficientemente derivables, calculamos sus derivadas segundas:
d2f1 = sp.diff(f1,x,2)
d2f2 = sp.diff(f2,x,2)
d2f3 = sp.diff(f3,x,2)

print('Derivada segunda en el primer tramo = ',d2f1) # -2 <0 => f cóncava en (-oo,-1)
print('Derivada segunda en el segundo tramo = ',d2f2)
print('Derivada segunda en el tercer tramo = ',d2f3) # > 0 => f convexa en (1,oo)
```

```python
# Estudiamos el signo de f2 entre -1 y 1. Para ello, vemos si la derivada segunda se anula en algún punto del intervalo:
print(sp.solve(d2f2,x)) # Dos raíces: 0 y 1
```

```python
# Evaluamos en un punto entre -1 y 0:
print('Valor de la derivada segunda en -1/2 = ',d2f2.subs(x,-0.5)) # >0 => f convexa en (-1,0)
# y en un punto entre 0 y 1:
print('Valor de la derivada segunda en 0.5 = ',d2f2.subs(x,0.5)) # <0 => f cóncava en (0,1)
```

```python
# Concluimos que -1, 0, 1 son puntos de inflexión
```
