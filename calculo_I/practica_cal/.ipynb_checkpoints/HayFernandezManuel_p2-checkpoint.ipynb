```python
# Ejercicio 1
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
x = sp.Symbol('x',real=True)
f = (x - 3)*(4-x)**2/3
# 1.Calcula su dominio
print('La función no tiene puntos problemáticos en ',sp.solve(f))
print(f.subs(x,3))
print(f.subs(x,4))
print('Pero en esos puntos existe imagen que es 0')
print('El dominio de la función es todo R')
# 2. Extremos relativos de f
df = sp.diff(f, x)
ext_rel = sp.solve(df, x)
print(f"Los extremos relativos son:",ext_rel)

# 3. Extremo absoluto en [3, 6]

print("f(3):", f.subs(x, 3))
print("f(4):",f.subs(x, 4))
print("f(5):",f.subs(x, 5))
print("f(6):",f.subs(x, 6))

# Encontrar el máximo absoluto
print("Los extremos absolutos son absoluto en [3, 6] son 4 que se alcanza en x = 6 y 0 que se alcanza en x=3 y x=4 ")
```

```python
# Ejercicio 2 
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# Definimos las variables 
x, a = sp.symbols('x a')

# Ramas de la función
f_izq = a*x**2 + 2*x - 1
f_der = 3*x + a - 2

# a. Estudio de la continuidad en x=1 
lim_izq = f_izq.subs(x, 1)
lim_der = f_der.subs(x, 1)
print(lim_izq)
print(lim_der)
# Ecuación de continuidad: lim_izq = lim_der
# a + 1 = a + 1
ecua_continuidad = 0
print("Ecuación de continuidad en x=1:", ecua_continuidad)
print(lim_izq)
print(lim_der)
# a+1 = a+1, por lo tanto a es continua siempre que sea un número real.
sol_a_cont = sp.solve(ecua_continuidad, a)
print(sol_a_cont)
print("a es continua siempre que sea un número real")


# b. Estudio de la derivabilidad en x=1 
df_izq = sp.diff(f_izq, x)
df_der = sp.diff(f_der, x)

# Evaluamos las derivadas en x=1
deriv_izq = df_izq.subs(x, 1)
deriv_der = df_der.subs(x, 1)

# Ecuación de derivabilidad
print(deriv_izq)
print(deriv_der)
# 2*a + 2 = 3
eq_derivabilidad = 2*a + 2 - 3
print("\nEcuación de derivabilidad en x=1:", eq_derivabilidad)

# Resolvemos para a
sol_a_deriv = sp.solve(eq_derivabilidad, a)
print("Solución para a para que sea derivable:", sol_a_deriv)
```
