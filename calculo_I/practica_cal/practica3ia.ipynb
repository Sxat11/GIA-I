## 1.1 Límites

En este apartado aprenderemos a calcular límites de funciones con **Python** empleando la librería **SymPy**.

Para calcular el límite de la expresión simbólica $x^2$ cuando la variable independiente $x$ se aproxima a una valor $a$, escribimos:

```python
import sympy as sp   # limite
# Definimos la variable simbólica x 
x = sp.Symbol('x')
# Definimos la expresión simbólica
f = x**2
# Definimos el punto en el que se calcula el límite
a = -3 
# Calculamos el límite de la expresión cuando x se aproxima al punto a
l1 = sp.limit(f, x, a) 
# Mostramos por pantalla el valor del límite
display(l1) 
```

### Limites laterales

También es posible calcular los límites laterales, añadiendo en la instruccion 'limit' un '+' para el límite por la derecha y un '-' para el límite por la izquierda. 

Para ilustrar este cálculo, usamos la función $f_1(x)=\dfrac{1}{x}$.

```python
# Definimos la expresión simbólica de la que vamos a calcular el límite
f1 = 1/x
# Límite de f1 en a=0 por la derecha
l2d = sp.limit(f1, x, 0,'+') 
print('Limite por la derecha = ',l2d)
# Límite de f1 en a=0 por la izquierda
l2i = sp.limit(f1, x, 0,'-')  
print('Limite por la izquierda = ',l2i)

# Observa el cálculo si no se indica el lado
av = sp.limit(f1, x, 0)  
display(av)
```

### Limites en el infinito.

También es posible calcular límites en el infinito:

```python
# Límite de f1 cuando x tiende a -sp.oo 
ahi = sp.limit(f1, x, -sp.oo) 
# Límite de f1 cuando x tiende a sp.oo
ahd = sp.limit(f1, x, sp.oo)  

# Mostramos los resultados por pantalla
display('Límite cuando x-> -sp.oo es ',ahi) 
display('Límite cuando x-> +sp.oo es ',ahd)
```

Calculamos ahora las asíntotas de la función $f(x)=\displaystyle\frac{x^2}{x+1}$. Podemos denominar a cada límite de modo que nos indique lo que se guarda en la variable correspondiente.

```python
# Definimos la expresión simbólica
f2 = x*x/(x + 1)


# ASINTOTAS HORIZONTALES
ahd = sp.limit(f2, x, sp.oo)
ahi = sp.limit(f2, x, -sp.oo)
print('Límite en +sp.oo =',ahd)
print('Límite en -sp.oo =',ahi)
# NO HAY ASINTOTAS HORIZONTALES


# ASINTOTAS VERTICALES
avd = sp.limit(f2, x, -1,'+')
avi = sp.limit(f2, x, -1,'-')
print('Limite en a=-1 por la derecha = ',avd)
print('Limite en a=-1 por la izquierda = ',avi)
# x = -1 asíntota vertical


# ASINTOTAS OBLICUAS
# Pendiente de la asíntota oblicua por la derecha
aomd = sp.limit(f2/x, x, sp.oo)         
print('aomd = ',aomd)
# y = aomd * x + aond: asíntota oblicua por la derecha
aond = sp.limit(f2 - aomd*x, x, sp.oo)  
print('aond = ',aond) 
# y = 1 * x - 1 = x-1 asíntota oblicua cuando x->+sp.oo

# Pendiente de la asíntota oblicua por la izquierda
aomi = sp.limit(f2/x, x, -sp.oo)      
print('aomi = ',aomi)
# y = aomi * x + aon: asíntota oblicua por la izquierda
aoni = sp.limit(f2 - aomi*x, x, -sp.oo)
print('aoni = ',aoni)
# y = 1 * x - 1 = x-1 asíntota oblicua cuando x->-sp.oo
```

### Ejercicio 2 

Representa en la misma ventana gráfica la función $f(x)=\displaystyle\frac{x^2}{x+1}$ y sus asíntotas.

```python
import sympy as sp
import matplotlib.pyplot as plt
import numpy as np

x_sym = sp.Symbol('x')
f_sym = (x_sym**2) / (x_sym + 1)

print("--- ANÁLISIS DE ASÍNTOTAS ---")

av = sp.limit(f_sym, x_sym, -1, dir='+-')
print(f"Límite cuando x tiende a -1: {av}")

m = sp.limit(f_sym / x_sym, x_sym, sp.oo)
n = sp.limit(f_sym - m * x_sym, x_sym, sp.oo)
print(f"Pendiente de la asíntota (m): {m}")
print(f"Ordenada al origen (m): {n}")
print(f"Asíntota oblicua: y = {m}*x + {n}")

print("\n--- REPRESENTACIÓN GRÁFICA ---")

x_vals_1 = np.linspace(-10, -1.1, 400)
x_vals_2 = np.linspace(-0.9, 10, 400)

f_num = lambda x: (x**2) / (x + 1)

y_vals_1 = f_num(x_vals_1)
y_vals_2 = f_num(x_vals_2)

plt.figure(figsize=(8, 6))
plt.plot(x_vals_1, y_vals_1, label='$f(x) = \\frac{x^2}{x+1}$', color='blue')
plt.plot(x_vals_2, y_vals_2, color='blue')

plt.axvline(x=-1, color='red', linestyle='--', label='Asíntota vertical: $x = -1$')

asymptote_oblique = x_vals_2 - 1 
plt.plot(x_vals_2, x_vals_2 - 1, color='green', linestyle='--', label='Asíntota oblicua: $y = x - 1$')

plt.xlim(-10, 10)
plt.ylim(-15, 15)
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Representación de $f(x) = \\frac{x^2}{x+1}$ y sus asíntotas')
plt.legend()
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```

### Ejercicio 3

Calcula todas las asíntotas de las funciones siguientes: 

 a)   $ g(t) = \displaystyle\frac{2t^3}{t^2 - 1} $

 b)   $ a(v) = 2 + \displaystyle \frac{3}{v^2 + 4} $
 
y represéntalas gráficamente. 

```python
import numpy as np
import matplotlib.pyplot as plt

t1 = np.linspace(-4, -1.05, 300)
t2 = np.linspace(-0.95, 0.95, 300)
t3 = np.linspace(1.05, 4, 300)

g = lambda t: (2*t**3) / (t**2 - 1)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(t1, g(t1), color='blue')
plt.plot(t2, g(t2), color='blue', label='$g(t) = \\frac{2t^3}{t^2-1}$')
plt.plot(t3, g(t3), color='blue')
# Asíntotas verticales
plt.axvline(x=1, color='red', linestyle='--', label='AV: $t = 1$')
plt.axvline(x=-1, color='red', linestyle=':', label='AV: $t = -1$')
# Asíntota oblicua
plt.plot(np.linspace(-4, 4, 100), 2*np.linspace(-4, 4, 100), color='green', linestyle='--', label='AO: $y = 2t$')
plt.title('Función A y sus asíntotas')
plt.xlabel('t')
plt.ylabel('g(t)')
plt.ylim(-15, 15)
plt.xlim(-4, 4)
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)
plt.legend(fontsize=8)
plt.grid(True, linestyle=':', alpha=0.6)

v = np.linspace(-10, 10, 400)
a = lambda v: 2 + 3 / (v**2 + 4)

plt.subplot(1, 2, 2)
plt.plot(v, a(v), color='purple', label='$a(v) = 2 + \\frac{3}{v^2+4}$')

plt.axhline(y=2, color='orange', linestyle='--', label='AH: $y = 2$')
plt.title('Función B y sus asíntotas')
plt.xlabel('v')
plt.ylabel('a(v)')
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)
plt.legend(fontsize=8)
plt.grid(True, linestyle=':', alpha=0.6)

plt.tight_layout()
plt.show()
print('No tiene oblicua')
```
